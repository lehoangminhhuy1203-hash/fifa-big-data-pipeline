# 00 – Kiểm tra kết nối cụm
Chạy lần lượt từng ô (Shift+Enter). Nếu cả 3 ô đều chạy không lỗi, cụm HDFS + Hive + Spark đã sẵn sàng.

```python
from pyspark.sql import SparkSession

spark = (SparkSession.builder.appName("kiem-tra-ket-noi")
         .config("spark.hadoop.fs.defaultFS", "hdfs://namenode:8020")
         .config("spark.sql.warehouse.dir", "hdfs://namenode:8020/user/hive/warehouse")
         .config("hive.metastore.uris", "thrift://hive-metastore:9083")
         .config("spark.sql.parquet.writeLegacyFormat", "true")
         .config("spark.driver.memory", "2g")
         .enableHiveSupport()
         .getOrCreate())
print("Spark version:", spark.version)
```

```python
# Test 1: ghi & đọc file Parquet trên HDFS
spark.range(1000).write.mode("overwrite").parquet("/tmp/test_parquet")
print("Đọc lại từ HDFS:", spark.read.parquet("/tmp/test_parquet").count(), "dòng")
```

```python
# Test 2: kết nối Hive Metastore
spark.sql("CREATE DATABASE IF NOT EXISTS fifa")
spark.sql("SHOW DATABASES").show()
```
